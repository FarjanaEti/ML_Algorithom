## Creating my own insurance csv

```python
import pandas as pd
import numpy as np
```

```python
age=np.random.randint(20,80, size=30)
```

```python
insurance=np.random.randint(0,2, size=30)
```

```python
df=pd.DataFrame({
    "Age":age,
    "Take_Insurance":insurance
})

df.to_csv('Insurance.csv', index=False)
```

```python
df
```

```python
## logistic regression

import matplotlib.pyplot as plt
#%matplotlib inline # display matplotlib charts directly inside the notebook
```

```python
df.head(5)
```

```python
plt.scatter(df.Age,df.Take_Insurance, marker='.', color='green')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
# shift+ tab
X_train,X_test,y_train,y_test=train_test_split(df[['Age']], df.Take_Insurance, test_size=0.1)
```

```python
X_test
```

```python
X_train
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.predict(X_test)
```

```python
model.predict(X_train)
```

```python
model.score(X_test,y_test)
```

```python
model.predict_proba(X_test)
```

```python
model.intercept_
```

```python
model.coef_
```

```python
import math
def sigmoid(x):
    return 1/(1+math.exp(-x))
```
