## Train test split

```python
from sklearn.model_selection import train_test_split
```

```python
X= df.drop(['target','flower_name'], axis='columns')
#all the numeric columns except target & flower_name
y=df.target
```

```python
X_train, X_test, y_train, y_test=train_test_split(X,y, test_size=0.2, random_state=1)
```

```python
len(X_train)
```

```python
#len(X_test)
len(y_test)
```

## Create KNN (K Neighrest Neighbour Classifier)

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier(n_neighbors=10)
```

```python
knn.fit(X_train,y_train)
```

```python
knn.score(X_test,y_test)
```

```python
knn.predict([[4.8,3.0,1.5,0.3]])
```

## Plot Confusion Matrix 

```python
## confusion metrix tell wich cells gives the right prediction and wich cell did't get the prediction right
```

```python
from sklearn.metrics import confusion_matrix
y_predict=knn.predict(X_test) #Predict using your trained mode

#confision_matrix=confusion_matrix(truth, prediction)
cm = confusion_matrix(y_test, y_predict)
cm

#Each row represents the actual labels (truth)
#Each column represents the predicted labels
#Diagonal values → correctly predicted examples
#Off-diagonal values → misclassified examples
# 12 predict correct 1 mistake 
```

```python
#data visualization library in Python, built on top of Matplotlib
#heatmap is a grid-like visualization that uses colors to represent numerical values.
# annot=you remove it, you’ll only see colors (no text labels).
import seaborn as sn
plt.figure(figsize=(7,5))
sn.heatmap(cm,annot=True)
plt.xlabel('Prediction')
plt.xlabel('Truth')
```

## Print classification report for precesion, recall and f1-score for each classes

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_predict))
```
