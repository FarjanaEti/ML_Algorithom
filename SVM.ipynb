```python
from sklearn.datasets import load_iris
import pandas as pd
data=load_iris()
```

```python
data.feature_names
```

```python
data.target_names
```

```python
df=pd.DataFrame(data.data, columns=data.feature_names)
df.head()
```

```python
df['target']=data.target
```

```python
df[df.target==1]
```

```python
df[df.target==2]
```

```python
# new col
df['flower_name']=df.target.apply(lambda x: data.target_names[x])
df.head(3)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X=df.drop(['target', 'flower_name'], axis='columns')
y=df.target
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.svm import SVC
#model=SVC(C=10) C=regularization in creasing c decresing score
#model=SVC(gamma=100) worst
#model=SVC(kernel='linear') to go kernel type shift+tab or kernel=rbf
model=SVC()
```

```python
model.fit(X_train,y_train)
## this give class sklearn.svm.SVC(*, C=1.0, kernel='rbf', degree=3, gamma='scale',
#coef0=0.0, shrinking=True, probability=False, tol=0.001, cache_size=200,
#class_weight=None, verbose=False, max_iter=-1, decision_function_shape='ovr', 
#break_ties=False, random_state=None)[source]
```

## 1. Regularization (C)

```python
#C controls how strict or flexible
#c=1 flexible,Allows more misclassifications to achieve a smoother decision boundary
#c=100 strict Tries to classify every training point correctly (may overfit)
#model_c=SVC(C=10)
model_C=SVC(C=1)
model_C.fit(X_train,y_train)
model.score(X_test, y_test)
```
