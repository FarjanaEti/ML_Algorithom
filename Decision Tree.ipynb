```python
import pandas as pd
```

```python
df=pd.read_csv('salaries.csv')
```

```python
df.head(5)
```

```python
input_data=df.drop('salary_more_then_100k', axis='columns')
```

```python
target=df['salary_more_then_100k']
```

```python
#LabelEncoder converts text labels (strings) into numbers
from sklearn.preprocessing import LabelEncoder
```

```python
## three diffrent object for three column
le_company=LabelEncoder()
le_job=LabelEncoder()
le_degree=LabelEncoder()
```

```python
# fit()-> learns all uniqe categpries
#transform()->convert into umeric code 
input_data['new_com']=le_company.fit_transform(input_data['company'])
input_data['new_job']=le_job.fit_transform(input_data['job'])
input_data['new_degree']=le_degree.fit_transform(input_data['degree'])
```

```python
input_data
## Number is decided alphabetically like apple=0 dell=1
```

```python
numeric_input=input_data.drop(['company','job','degree'], axis='columns')
```

```python
numeric_input.head(4)
```

```python
target
```

```python
from sklearn import tree
model=tree.DecisionTreeClassifier()
```

```python
model.fit(numeric_input, target)
```

```python
model.score(numeric_input, target)
```

# Is salary of Google, Computer Engineer, Bachelors degree > 100 k ?

```python
# it is from input_data com-job-degree
# array[0] means he won't get the job
model.predict([[2,1,0]])
```
